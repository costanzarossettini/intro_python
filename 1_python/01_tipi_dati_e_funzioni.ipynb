# 1. Che cos'è un tipo di dato?

Una variabile può contenere valori diversi.

```python
nome = "Anna"
eta = 25
altezza = 1.68
```

Questi valori non sono tutti dello stesso tipo:

- `"Anna"` è un testo;
- `25` è un numero intero;
- `1.68` è un numero decimale.

Il **tipo di dato** indica che cosa rappresenta un valore e quali operazioni possiamo eseguire su di esso.

## Tutti i tipi built-in principali

| Categoria | Tipi |
|---|---|

| Numeri | `int`, `float`, `complex` |
| Sequenze | `list`, `tuple`, `range` |
| Associazioni chiave-valore | `dict` |
| Insiemi | `set`, `frozenset` |
| Valori logici | `bool` |
| Dati binari | `bytes`, `bytearray`, `memoryview` |
| Assenza di valore | `NoneType` |


## La funzione `type()`

La funzione built-in `type()` restituisce il tipo di un valore.

````python
# Panoramica rapida di diversi tipi

testo = "Python"
intero = 10
decimale = 3.5
complesso = 2 + 3j
sequenza_modificabile = [1, 2, 3]
sequenza_fissa = (1, 2, 3)
intervallo = range(5)
persona = {"nome": "Anna", "eta": 25}
numeri_unici = {1, 2, 3}
vero_o_falso = True
nessun_valore = None

print(type(testo))
print(type(intero))
print(type(decimale))
print(type(complesso))
print(type(sequenza_modificabile))
print(type(sequenza_fissa))
print(type(intervallo))
print(type(persona))
print(type(numeri_unici))
print(type(vero_o_falso))
print(type(nessun_valore))
````

L'output contiene espressioni come:

```text
<class 'str'>
<class 'int'>
<class 'float'>
```

La parola `class` indica che in Python ogni valore appartiene a una classe, cioè a un tipo.

# 2. Il tipo `str`: il testo

https://docs.python.org/3/library/stdtypes.html#str search class str 

Una stringa è una sequenza di caratteri racchiusa tra apici singoli o doppi.

```python
nome = "Luca"
citta = 'Roma'
```

Possiamo:

- unire stringhe con `+`;
- ripetere una stringa con `*`;
- misurarne la lunghezza con `len()`;
- leggere un carattere tramite il suo indice.

````python
nome = "Luca"
cognome = "Rossi"

nome_completo = nome + " " + cognome

print(nome_completo)
print("Numero di caratteri:", len(nome_completo))
print("Primo carattere:", nome_completo[0])
print("Testo ripetuto:", "Ciao! " * 3)
````

Gli indici iniziano da `0`.

Nella stringa `"Python"`:

| Carattere | P | y | t | h | o | n |
|---|---:|---:|---:|---:|---:|---:|
| Indice | 0 | 1 | 2 | 3 | 4 | 5 |

````python
linguaggio = "Python"

print(linguaggio[0])
print(linguaggio[1])
print(linguaggio[-1])  # -1 indica l'ultimo carattere
````

 Principali metodi delle stringhe (`str`)

Le stringhe in Python sono **immutabili**: questi metodi restituiscono una nuova stringa senza modificare quella originale.

1. **`str.lower()`**
   Converte tutti i caratteri della stringa in minuscolo.

2. **`str.upper()`**
   Converte tutti i caratteri della stringa in maiuscolo.

3. **`str.capitalize()`**
   Rende maiuscolo il primo carattere e minuscoli tutti gli altri.

4. **`str.strip()`**
   Rimuove spazi e caratteri di nuova riga dall'inizio e dalla fine della stringa.

5. **`str.replace(old, new)`**
   Sostituisce tutte le occorrenze di `old` con `new`.

6. **`str.split(sep)`**
   Divide la stringa in una lista utilizzando `sep` come separatore.

7. **`str.join(iterable)`**
   Unisce più stringhe utilizzando la stringa su cui viene chiamato come separatore.

8. **`str.find(sub)`**
   Restituisce la posizione della prima occorrenza di `sub`, oppure `-1` se non viene trovata.

9. **`str.count(sub)`**
   Conta quante volte la sottostringa `sub` compare nella stringa.

10. **`str.startswith(prefix)`**
    Restituisce `True` se la stringa inizia con `prefix`, altrimenti restituisce `False`.

# 6. Le sequenze: `list`, `tuple` e `range`

https://docs.python.org/3/library/stdtypes.html#sequence-types-list-tuple-range

Una sequenza contiene più valori in un ordine preciso.

I tre tipi principali sono:

- `list`: sequenza modificabile;
- `tuple`: sequenza non modificabile;
- `range`: sequenza di numeri, spesso usata nei cicli.

## `list`: una sequenza modificabile

Una lista è racchiusa tra parentesi quadre.

```python
frutti = ["mela", "pera", "banana"]
```

Gli elementi possono essere letti, aggiunti, modificati o rimossi.

````python
frutti = ["mela", "pera", "banana"]

print(frutti)
print("Primo elemento:", frutti[0])
print("Numero di elementi:", len(frutti))

frutti.append("arancia")
print("Dopo append:", frutti)

frutti[1] = "kiwi"
print("Dopo la modifica:", frutti)

frutti.remove("banana")
print("Dopo remove:", frutti)

frutti.reverse()
print("Dopo reverse:", frutti)

frutti.pop()
print("Dopo pop:", frutti)
````

Alcuni strumenti comuni per le liste:

| Operazione | Significato |
|---|---|
| `lista[indice]` | legge un elemento |
| `len(lista)` | conta gli elementi |
| `lista.append(valore)` | aggiunge un elemento |
| `lista.remove(valore)` | rimuove un elemento |
| `lista[indice] = valore` | modifica un elemento |

## `tuple`: una sequenza non modificabile

Una tupla è racchiusa tra parentesi tonde.

```python
coordinate = (10, 20)
```

Dopo la creazione, i suoi elementi non possono essere modificati.

````python
coordinate = (10, 20)

print(coordinate)
print("Coordinata x:", coordinate[0])
print("Coordinata y:", coordinate[1])
print("Numero di elementi:", len(coordinate))
````

Questa istruzione produrrebbe un errore:

```python
coordinate[0] = 50
```

Una tupla è utile quando i dati devono restare fissi.

## `range`: una sequenza di numeri

`range()` genera una sequenza di numeri.

È usata spesso insieme al ciclo `for`.

````python
numeri = range(1, 6)

print(numeri)
print(list(numeri))

for numero in range(1, 6):
    print(numero)
````

Il limite finale è escluso.

```python
range(1, 6)
```

produce:

```text
1, 2, 3, 4, 5
```

# 7. Il tipo `dict`

https://docs.python.org/3/library/stdtypes.html#mapping-types-dict

Un dizionario associa una **chiave** a un **valore**.

```python
studente = {
    "nome": "Anna",
    "eta": 22,
    "corso": "Python"
}
```

In una lista usiamo un indice numerico.  
In un dizionario usiamo una chiave.

````python
studente = {
    "nome": "Anna",
    "eta": 22,
    "corso": "Python"
}

print(studente)
print("Nome:", studente["nome"])
print("Età:", studente["eta"])

studente["eta"] = 23
studente["citta"] = "Milano"

print("Dizionario aggiornato:", studente)

studente.pop("corso")
print("Dizionario aggiornato senza corso:", studente)
````

Alcuni strumenti comuni per i dizionari:

| Operazione | Significato |
|---|---|
| `dizionario[chiave]` | legge il valore |
| `dizionario[chiave] = valore` | aggiunge o modifica |
| `len(dizionario)` | conta le coppie |
| `dizionario.keys()` | restituisce le chiavi |
| `dizionario.values()` | restituisce i valori |
| `dizionario.items()` | restituisce coppie chiave-valore |

````python
voti = {
    "italiano": 8,
    "matematica": 7,
    "inglese": 9
}

print("Materie:", list(voti.keys()))
print("Voti:", list(voti.values()))
print("Coppie:", list(voti.items()))
````

````python
# you can also build a dictionary from lists
keys = ["name", "age", "city"]
values = ["Bob", 25, "London"]

result = dict(zip(keys, values))
print(result)
````

# 8. `set` e `frozenset`

## `set`

Un set è una raccolta:

- senza elementi duplicati;
- senza una posizione fissa garantita;
- modificabile.

È utile per eliminare duplicati o controllare appartenenze.

````python
colori = {"rosso", "blu", "rosso", "verde", "blu"}

print(colori)
print("Numero di colori diversi:", len(colori))

colori.add("giallo")
print(colori)
````

Possiamo convertire una lista in un set per eliminare i duplicati.

````python
nomi = ["Anna", "Luca", "Anna", "Marco", "Luca"]

nomi_unici = set(nomi)

print("Lista originale:", nomi)
print("Nomi senza duplicati:", nomi_unici)
````

## `frozenset`

Un `frozenset` è simile a un set, ma non può essere modificato dopo la creazione.

````python
giorni_festivi = frozenset(["sabato", "domenica"])

print(giorni_festivi)
print(type(giorni_festivi))
````

# 9. Tipi binari: `bytes`, `bytearray` e `memoryview`

Questi tipi sono usati per lavorare con dati binari, file, immagini, audio e comunicazioni di rete.

Per ora è sufficiente conoscerne la differenza generale:

- `bytes`: sequenza binaria non modificabile;
- `bytearray`: sequenza binaria modificabile;
- `memoryview`: vista su dati binari senza copiarli.

Sono tipi importanti, ma non sono normalmente necessari nei primi programmi.

````python
dati_fissi = bytes([65, 66, 67])
dati_modificabili = bytearray([65, 66, 67])
vista = memoryview(dati_modificabili)

print(dati_fissi)
print(dati_modificabili)
print(vista)
print(type(dati_fissi))
print(type(dati_modificabili))
print(type(vista))
````

I numeri `65`, `66` e `67` corrispondono ai caratteri `A`, `B` e `C` nella codifica ASCII.

# 13. Le funzioni

Una funzione è un blocco di codice con un nome.

Possiamo eseguirlo ogni volta che serve.

La forma più semplice è:

```python
def nome_funzione():
    istruzioni
```

Per eseguire la funzione dobbiamo chiamarla:

```python
nome_funzione()
```

````python
def saluta():
    print("Ciao!")

saluta()
saluta()
````

## Parametri

Un parametro permette alla funzione di ricevere un valore.

````python
def saluta_persona(nome):
    print(f"Ciao, {nome}!")

saluta_persona("Anna")
saluta_persona("Luca")
````

La funzione può avere più parametri.

````python
def presenta_persona(nome, eta):
    print(f"{nome} ha {eta} anni.")

presenta_persona("Anna", 25)
presenta_persona("Luca", 31)
````

## Il valore restituito con `return`

`return` invia un risultato fuori dalla funzione.

Il risultato può essere salvato in una variabile.

````python
def somma(a, b):
    risultato = a + b
    return risultato

totale = somma(10, 5)

print(totale)
````

Possiamo scrivere la stessa funzione in forma più breve.

````python
def somma(a, b):
    return a + b

print(somma(10, 5))
print(somma(3, 7))
````

## Differenza tra `print()` e `return`

- `print()` mostra un valore sullo schermo;
- `return` restituisce un valore che può essere riutilizzato.

Osserviamo due funzioni simili.

````python
def somma_con_print(a, b):
    print(a + b)

def somma_con_return(a, b):
    return a + b

risultato_1 = somma_con_print(2, 3)
risultato_2 = somma_con_return(2, 3)

print("Valore di risultato_1:", risultato_1)
print("Valore di risultato_2:", risultato_2)
````

`risultato_1` vale `None`, perché la prima funzione stampa il risultato ma non lo restituisce.

`risultato_2` vale `5`, perché la seconda funzione usa `return`.

## Parametri con un valore predefinito

Possiamo assegnare un valore predefinito a un parametro.

````python
def saluta(nome, messaggio="Ciao"):
    return f"{messaggio}, {nome}!"

print(saluta("Anna"))
print(saluta("Luca", "Buongiorno"))
````

Il parametro `messaggio` vale `"Ciao"` quando non viene fornito un altro valore.

## Esercizio svolto 5 — Area di un rettangolo

Scriviamo una funzione che riceve base e altezza e restituisce l'area.

## Esercizio svolto 6 — Media di tre numeri

Scriviamo una funzione che riceve tre numeri e restituisce la loro media.

````python
def media_tre_numeri(a, b, c):
    somma = a + b + c
    media = somma / 3
    return media

print(media_tre_numeri(6, 8, 10))
````

## Esercizio svolto 7 — Controllare se un numero è pari

Un numero è pari quando il resto della divisione per `2` è uguale a `0`.

Non è necessario usare `if`: il confronto produce direttamente un booleano.

````python
def numero_pari(numero):
    return numero % 2 == 0

print(numero_pari(8))
print(numero_pari(7))
````

## Esercizio svolto 8 — Riassunto di una lista

La funzione riceve una lista di numeri e restituisce un dizionario con alcune statistiche.

````python
def riassunto_numeri(numeri):
    risultato = {
        "quantita": len(numeri),
        "somma": sum(numeri),
        "minimo": min(numeri),
        "massimo": max(numeri)
    }

    return risultato

dati = riassunto_numeri([4, 8, 2, 9])

print(dati)
print("Somma:", dati["somma"])
print("Massimo:", dati["massimo"])
````

#### Esercizio 1 — Conversione tra basi

Converti il numero decimale `157` in:

- binario;
- ottale;
- esadecimale.

Python mette a disposizione le funzioni built-in:

- `bin()` per il binario;
- `oct()` per l'ottale;
- `hex()` per l'esadecimale.

````python
numero = 157

numero_binario = bin(numero)
numero_ottale = oct(numero)
numero_esadecimale = hex(numero)

print("Numero decimale:", numero)
print("Binario:", numero_binario)
print("Ottale:", numero_ottale)
print("Esadecimale:", numero_esadecimale)
````

### Spiegazione

I prefissi indicano la base utilizzata:

- `0b` indica il sistema binario;
- `0o` indica il sistema ottale;
- `0x` indica il sistema esadecimale.

Quindi:

```text
157₁₀ = 10011101₂ = 235₈ = 9d₁₆
```

Per rimuovere il prefisso possiamo usare uno *slice*:

```python
bin(157)[2:]
```

````python
print("Binario senza prefisso:", bin(157)[2:])
print("Ottale senza prefisso:", oct(157)[2:])
print("Esadecimale senza prefisso:", hex(157)[2:])
````

#### Esercizio 2 — Confronto tra `0.1 + 0.2` e `0.3`

Confronta `0.1 + 0.2` con `0.3` e spiega il risultato con un commento.

````python
somma = 0.1 + 0.2
valore_atteso = 0.3

print("0.1 + 0.2 =", somma)
print("0.3       =", valore_atteso)
print("Sono esattamente uguali?", somma == valore_atteso)

# Il confronto restituisce False perché molti numeri decimali
# non possono essere rappresentati esattamente nel formato binario
# usato dai float. Il valore calcolato è leggermente maggiore di 0.3.
````

### Spiegazione

I numeri `float` vengono memorizzati in formato binario. Valori come `0.1`, `0.2` e `0.3` non hanno una rappresentazione binaria finita esatta.

Di conseguenza, Python memorizza approssimazioni molto vicine ai valori richiesti.

Per confrontare numeri decimali calcolati è spesso preferibile verificare che la loro differenza sia molto piccola.

````python
tolleranza = 1e-9
differenza = abs(somma - valore_atteso)

print("Differenza:", differenza)
print("Uguali entro la tolleranza?", differenza < tolleranza)
````

#### Esercizio 3 — Arrotondamento senza `round()`

Scrivi una funzione che arrotonda un numero a `n` cifre decimali senza usare `round()`.

La funzione seguente usa:

- una potenza di `10` per spostare temporaneamente la virgola;
- `math.floor()` per i numeri positivi;
- `math.ceil()` per i numeri negativi.

In caso di cifra esattamente a metà, arrotonda allontanandosi dallo zero.

````python
import math

def arrotonda(numero, cifre_decimali):
    fattore = 10 ** cifre_decimali
    numero_spostato = numero * fattore

    if numero_spostato >= 0:
        intero_arrotondato = math.floor(numero_spostato + 0.5)
    else:
        intero_arrotondato = math.ceil(numero_spostato - 0.5)

    return intero_arrotondato / fattore


print(arrotonda(3.14159, 2))
print(arrotonda(8.765, 2))
print(arrotonda(-2.345, 2))
print(arrotonda(125.6, 0))
````

### Come funziona

Per arrotondare `3.14159` a due cifre decimali:

1. calcoliamo `10 ** 2`, cioè `100`;
2. moltiplichiamo: `3.14159 * 100 = 314.159`;
3. aggiungiamo `0.5` e prendiamo la parte intera inferiore;
4. dividiamo nuovamente per `100`.

La distinzione tra numeri positivi e negativi evita che i valori negativi vengano arrotondati nella direzione sbagliata.

> Anche questa funzione lavora con valori `float`, quindi conserva i normali limiti di precisione dei numeri in virgola mobile.

#### Esercizio 4 — Errore assoluto e relativo

Calcola l'errore assoluto e relativo tra:

- valore vero: `42`;
- valore misurato: `41.6`.

Formule:

```text
errore assoluto = |valore vero - valore misurato|

errore relativo = errore assoluto / |valore vero|
```

````python
valore_vero = 42
valore_misurato = 41.6

errore_assoluto = abs(valore_vero - valore_misurato)
errore_relativo = errore_assoluto / abs(valore_vero)
errore_percentuale = errore_relativo * 100

print("Valore vero:", valore_vero)
print("Valore misurato:", valore_misurato)
print("Errore assoluto:", errore_assoluto)
print("Errore relativo:", errore_relativo)
print("Errore percentuale:", errore_percentuale, "%")
````

### Interpretazione

L'errore assoluto è espresso nella stessa unità di misura dei valori originali.

L'errore relativo non ha unità di misura e permette di valutare l'errore in rapporto alla grandezza del valore vero.

In questo caso la misura differisce dal valore vero di `0.4`, cioè di circa lo `0.95%`.

#### Esercizio 5 — Tabella delle potenze di 2

Genera una lista di tuple con le potenze di `2` da `2⁰` a `2¹⁶`.

Ogni tupla deve contenere:

```text
(esponente, risultato)
```

````python
tabella_potenze = []

for esponente in range(17):
    valore = 2 ** esponente
    tabella_potenze.append((esponente, valore))

print(tabella_potenze)
````


Possiamo attraversare la lista con un ciclo `for` e stampare una riga alla volta.

````python
for esponente, valore in tabella_potenze:
    print(f"2^{esponente:2} = {valore:5}")
````


La stessa lista può essere generata con una *list comprehension*:

```python
tabella_potenze = [(esponente, 2 ** esponente) for esponente in range(17)]
```

La forma con il ciclo esplicito è però più facile da leggere per chi sta iniziando.
