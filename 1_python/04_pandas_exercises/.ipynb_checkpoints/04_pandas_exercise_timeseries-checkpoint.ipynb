> **Suggerimento:** usa `.set_index('Date')`; ricorda di riassegnare il risultato ad `apple` oppure di specificare `inplace=True`.

### Passaggio 7 — Esistono date duplicate?

> **Suggerimento:** `apple.index.duplicated()` restituisce un booleano per ogni data. Aggiungi `.any()` per sapere se almeno una è duplicata, oppure `.sum()` per contarle.

### Passaggio 8 — Ordina l'indice dalla data più vecchia alla più recente

> **Suggerimento:** usa `.sort_index(ascending=True)`. Non affidarti all'ordine originale del file.

### Passaggio 10 — Quanti giorni separano la data più vecchia da quella più recente?

> **Suggerimento:** sottrai `apple.index.min()` da `apple.index.max()`. Il risultato è un `Timedelta`; usa `.days` per il numero intero di giorni.
