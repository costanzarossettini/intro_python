# Filtrare e ordinare i dati

In questo esercizio caricherai dati dal web, creerai filtri booleani e ordinerai un DataFrame.


### Passaggio 1 — Importa le librerie necessarie

> **Suggerimento:** importa pandas con l'alias `pd`.

### Passaggio 2 — Leggi il dataset

Importa il file da questo [indirizzo](https://raw.githubusercontent.com/justmarkham/DAT8/master/data/chipotle.tsv).

> **Attenzione:** è un file TSV, quindi usa `pd.read_csv(url, sep='\t')`.

### Passaggio 3 — Salva i dati nella variabile `chipo`

Esamina `.head()`, `.columns` e `.dtypes`: capire la struttura prima di filtrare evita molti errori.

### Passaggio 4 — Quanti prodotti costano più di 10 dollari?

`item_price` contiene il simbolo `$`, quindi inizialmente è testo. Prima rimuovi il simbolo e converti la colonna in numeri. Conta poi i **prodotti distinti**, non tutte le righe d'ordine.


### Passaggio 5 — Qual è il prezzo di ogni prodotto?

Mostra un DataFrame con le sole colonne `item_name`, `choice_description` e `item_price`. Ogni combinazione prodotto-descrizione-prezzo deve comparire una sola volta.

> **Suggerimento:** seleziona le colonne con una lista, poi usa `.drop_duplicates()` e, se vuoi, `.sort_values('item_name')`.

### Passaggio 6 — Ordina le righe per nome del prodotto

> **Suggerimento:** usa `chipo.sort_values(by='item_name')`. Il metodo restituisce un nuovo DataFrame se non specifichi `inplace=True`.

### Passaggio 7 — Qual è la quantità associata all'articolo più costoso ordinato?

> **Suggerimento avanzato:** trova l'indice del prezzo massimo con `chipo['item_price'].idxmax()`, poi seleziona la riga con `.loc[indice]` e la colonna `quantity`. `.loc` seleziona in base all'etichetta dell'indice.
