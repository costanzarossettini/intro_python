# Jupyter Notebook


#### Che cos'è e a cosa serve Jupyter notebook
Jupyter Notebook è un’applicazione che permette di creare documenti interattivi chiamati notebook.

Un file Jupyter Notebook ha estensione **.ipynb**. Al suo interno si possono alternare:

- celle di testo, scritte in Markdown;
- celle di codice, per esempio in Python o in R;
- output, cioè i risultati prodotti dal codice.

Consigli, trucchi e scorciatoie utili in Jupyter: https://www.dataquest.io/blog/jupyter-notebook-tips-tricks-shortcuts/

#### Come installare Jupyter Notebook su Windows

Guida generale: https://www.geeksforgeeks.org/installation-guide/install-jupyter-notebook-in-windows/ 

Anaconda windows: https://www.youtube.com/watch?v=hSo1b2ITwCM

<img src="./meme_indian_guy.png" width="200">

1. *Metodo consigliato: installazione con Anaconda*<br>
    - Anaconda è una **distribuzione di Python** pensata per rendere più semplice l’installazione e la gestione degli strumenti usati per programmare, analizzare dati e lavorare con Jupyter Notebook.<br>

    - Anaconda permette anche di creare e gestire gli **ambienti virtuali**, spazi separati in cui installare pacchetti specifici per un progetto, senza creare confusione con altri progetti.<br>
    <br>
2. *Metodo alternativo: installazione con pip* <br>
    - Per chi ha già Python installato e sa usare il terminale, Jupyter Notebook può essere installato anche con **pip**, il gestore di pacchetti di Python.
   
   
Dopo l'installazione Jupyter Notebook può essere aperto da Anaconda Navigator cliccando su “Launch”, oppure da Anaconda Prompt usando il comando ```jupyter notebook```.

# GitHub

Documentazione ufficiale di github: https://docs.github.com/en/get-started/start-your-journey/hello-world <br>


#### Che cos’è GitHub

GitHub è una piattaforma online usata per salvare, condividere e gestire progetti digitali. <br>

La sua funzione principale è il **controllo di versione**, cioè la possibilità di tenere traccia delle modifiche fatte a un file nel tempo.

In pratica, invece di avere file chiamati:

            documento.doc
            documento_finale.doc
            documento_finale_definitivo.doc

GitHub permette di conservare tutte le versioni in modo ordinato e recuperabile.

#### A cosa serve GitHub

- salvare progetti online;
- tenere traccia delle modifiche;
- collaborare con altre persone;
- condividere codice, documenti o guide;
- proporre modifiche senza alterare direttamente il lavoro originale;
- consultare versioni precedenti di un progetto.

#### Concetti principali

1. **Repository**

    Un repository è lo spazio in cui viene salvato un progetto.

    Può contenere:

        file;
        cartelle;
        codice;
        immagini;
        documenti;
        guide.

    È come una cartella online dedicata a un progetto.<br>

<br>

2. **README**

    Il file README.md è di solito il primo file che si vede quando si apre un repository.

    Serve per spiegare:

    di cosa parla il progetto;
    come usarlo;
    quali sono le informazioni principali;
    eventuali istruzioni per installazione o utilizzo.

    Il README è scritto spesso in Markdown, un linguaggio semplice per formattare testi.<br>
<br>

3. **Issue**

    Le Issue servono per segnalare problemi, proporre idee o discutere modifiche.

    Possono essere usate per:

        indicare errori;
        proporre miglioramenti;
        organizzare attività;
        discutere aspetti del progetto.
        Branch 
<br>

4. **Pull Request**

    Una Pull Request è una proposta di modifica.

    Dopo aver lavorato su un branch, si può aprire una pull request per chiedere che le modifiche vengano controllate e, se approvate, inserite nel progetto principale.<br>

<br>

5. **Fork**

    Un fork è una copia di un repository sul proprio profilo GitHub.

    Serve per modificare liberamente un progetto senza cambiare quello originale.

    È utile quando si vuole partire da un progetto esistente e adattarlo alle proprie esigenze.<br>

<br>

6. **Funzioni social di GitHub**

    GitHub ha anche alcune funzioni simili a quelle di un social network:

        Follow: seguire un utente;
        Star: salvare un repository tra i preferiti;
        Watch: ricevere notifiche sugli aggiornamenti di un repository.


# Git

Comandi di git: https://education.github.com/git-cheat-sheet-education.pdf


A proposito di controllo di versione.. **git** è un software open che tiene traccia di tutte le modifiche apportate ai file di un progetto. Permette di registrare le varie versioni di un codice, fare esperimenti in rami paralleli e tornare indietro nel tempo in caso di errori, senza aver bisogno di una connessione Internet.

Git, essendo un pacchetto che gira in locale, va installato: https://git-scm.com/book/en/v2/Getting-Started-Installing-Git, https://github.com/git-guides/install-git. 

#### Principali comandi di git

| Comando                     | A cosa serve               |
| --------------------------- | -------------------------- |
| `git init`                  | Crea un repository Git     |
| `git clone URL`             | Scarica un repository      |
| `git status`                | Mostra lo stato dei file   |
| `git add .`                 | Prepara le modifiche       |
| `git commit -m "messaggio"` | Salva una versione         |
| `git log`                   | Mostra la cronologia       |
| `git branch`                | Mostra i branch            |
| `git checkout branch`       | Cambia branch              |
| `git merge branch`          | Unisce un branch           |
| `git pull`                  | Scarica aggiornamenti      |
| `git push origin main`      | Carica modifiche su GitHub |

Per maggiori informazioni:
