## 1. Sintassi base

In C++ il codice usa spesso:
- `;` alla fine delle istruzioni
- `{}` per delimitare i blocchi

In Python invece:
- non si usa `;` normalmente
- i blocchi si aprono con `:`
- il rientro (indentazione) è parte della sintassi

### Confronto 

**C++**
```cpp
int x = 5;
if (x > 0) {
    cout << "positivo";
}
```

**Python**
```python
x = 5
if x > 0:
    print("positivo")
```


````python
x = 5
if x > 0:
    print("positivo")
````

## 2. Commenti

I commenti servono a spiegare il codice.

### Confronto 


**C++**
```cpp
// commento su una riga

/* commento
   su più righe */
```

**Python**
```python
# commento su una riga
```

In Python, per spiegazioni più lunghe si usa spesso una sequenza di commenti `#`, oppure stringhe multilinea in contesti specifici.


````python
# Questo è un commento
nome = "Python"
print("Sto studiando", nome)
````

## 3. Variabili e tipi

In C++ di solito bisogna dichiarare il tipo:

```cpp
int eta = 20;
double prezzo = 9.99;
string nome = "Luca";
```

In Python il tipo viene gestito automaticamente:

```python
eta = 20
prezzo = 9.99
nome = "Luca"
```

Questo rende Python più rapido da scrivere nelle fasi iniziali.


````python
eta = 20
prezzo = 9.99
nome = "Luca"

# python capisce da solo il tipo della variabile 
print(eta, type(eta))
print(prezzo, type(prezzo))
print(nome, type(nome))
````

## 4. Conversione di tipo (cast)

Quando serve convertire un valore:

**C++**
```cpp
int x = (int)3.8;
```

**Python**
```python
x = int(3.8)
```

Funzioni utili in Python:
- `int()`
- `float()`
- `str()`
- `bool()`


````python
numero_decimale = 3.8
numero_intero = int(numero_decimale)
testo = str(numero_intero)

print(numero_decimale)
print(numero_intero)
print(testo)
print(type(testo))
````

## 5. Liste al posto degli array semplici

In Python si usano molto spesso le **liste**.

**C++**
```cpp
int numeri[3] = {10, 20, 30};
```

**Python**
```python
numeri = [10, 20, 30]
```

Anche l'accesso agli elementi parte da indice `0` in entrambi i linguaggi.


````python
numeri = [10, 20, 30]

print(numeri)
print(numeri[0])
print(numeri[1])
print(numeri[2])
````

## 6. Operazioni base sulle liste

Con le liste in Python puoi:
- leggere un elemento
- modificarlo
- aggiungerne uno con `append()`

Questo è molto comodo rispetto agli array classici di C++.


````python
numeri = [10, 20, 30]
numeri[1] = 25
numeri.append(40)

print(numeri)
````

## 7. Operatori di confronto e logici

Prima di usare bene `if`, serve conoscere gli operatori principali.

### Confronto
- `==` uguale
- `!=` diverso
- `>` maggiore
- `<` minore
- `>=` maggiore o uguale
- `<=` minore o uguale

### Logici

**C++**
```cpp
&&   ||   !
```

**Python**
```python
and  or   not
```


````python
a = 10
b = 5

print(a > b)
print(a == b)
print(a > 0 and b > 0)
print(a < 0 or b > 0)
print(not (a == b))
````

## 8. If

La struttura `if` serve per eseguire codice solo se una condizione è vera.

**C++**
```cpp
if (x > 0) {
    cout << "positivo";
}
```

**Python**
```python
if x > 0:
    print("positivo")
```


````python
x = 7

if x > 0:
    print("Il numero è positivo")
````

## 9. If / else / elif

In Python:
- `if` controlla la prima condizione
- `elif` aggiunge altri casi
- `else` gestisce il caso finale

**Python**
```python
if voto >= 18:
    print("Promosso")
else:
    print("Bocciato")
```


````python
voto = 24

if voto >= 30:
    print("Ottimo")
elif voto >= 18:
    print("Promosso")
else:
    print("Bocciato")
````

## 10. While

Il ciclo `while` ripete un blocco finché una condizione è vera.

**C++**
```cpp
while (i < 5) {
    cout << i << endl;
    i++;
}
```

**Python**
```python
while i < 5:
    print(i)
    i += 1
```

Fun fact: in c++ esiste una struttura `do while` che esegue il blocco almeno una volta, poi controlla la condizione. In python non esiste alcuna struttura con la stessa funzione.

````python
i = 0

while i < 5:
    print(i)
    i += 1
````

## 11. For con `range()`

Questo è uno dei punti in cui Python cambia di più rispetto a C++.

**C++**
```cpp
for (int i = 0; i < 5; i++) {
    cout << i << endl;
}
```

**Python**
```python
for i in range(5):
    print(i)
```

`range(5)` genera i numeri da `0` a `4`.


````python
for i in range(5):
    print(i)
````

## 12. For su una lista

In Python spesso non serve usare l'indice: puoi ciclare direttamente sugli elementi.

**C++**
```cpp
for (int i = 0; i < 3; i++) {
    cout << numeri[i] << endl;
}
```

**Python**
```python
for numero in numeri:
    print(numero)
```

Questo stile è molto "pythonico" ed è bene impararlo presto.


````python
numeri = [10, 20, 30]

for numero in numeri:
    print(numero)
````

## 13.  Riassunto 

| Argomento | C++ | Python | Nota pratica |
|---|---|---|---|
| Sintassi dei blocchi | `{}` | `:` + indentazione | In Python il rientro è obbligatorio |
| Fine istruzione | `;` | nessun `;` di solito | Codice più pulito visivamente |
| Commenti | `//`, `/* */` | `#` | Più semplice nelle basi |
| Variabili | tipo esplicito | tipo gestito automaticamente | Più veloce per iniziare |
| Cast | `(int)x` | `int(x)` | Molto leggibile |
| Array / liste | array, `vector` | liste `[]` | Le liste sono molto usate |
| Accesso elementi | `a[0]` | `a[0]` | Uguale come idea |
| Operatori logici | `&&`, `||`, `!` | `and`, `or`, `not` | Più vicini al linguaggio naturale |
| If | `if (...) {}` | `if ...:` | Meno simboli |
| While | `while (...) {}` | `while ...:` | Struttura simile |
| For con indice | `for (int i=0; ... )` | `for i in range(...)` | Python cambia parecchio qui |
| For su collezioni | spesso con indice | `for x in lista:` | Più diretto e leggibile |
