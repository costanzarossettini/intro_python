# Esercizio con git e github

Obiettivo: fare una piccola modifica a una repository GitHub. <br>
Repository di partenza: <br>
https://github.com/giacomo-b/rust-stakeholder

1. Fai un fork della repository
    1. Apri la repository su GitHub.
    2. Clicca su Fork.
    3. Crea una copia della repository nel tuo account GitHub. <br>
<br>

2. Clona il tuo fork in locale (così da poterlo modificare)

    1. github -> code -> HTTPS e poi da terminale sul tuo pc: ```git clone https://github.com/USERNAME/rust-stakeholder.git ```
    
    2. Da terminale, spostati nella nuova cartella con: ```cd rust-stakeholder ``` <br>
<br>    
 
3. Modifica la repo  <br>
Da locale, modifica un file qualisiasi della repo, ad esempio: apri il file README.md e cerca la sezione con gli esempi di utilizzo e aggiungi un nuovo esempio.
<br>

4. Controlla cosa è cambiato <br>
```
git status
git diff
```

5. Carica le modifiche appena apportate su github <br>
```
git add README.md
git commit -m "Add new README usage example"
git push origin main
```
Controlla che si sia aggiornato anche su github.

6. Gestisci gitignore <br>
    1. Crea un nuovo file txt da locale e salvalo dentro la repo. Questo file non sarà tracciato da git e dunque sarà ignorato durante i `push` request. 
    2. Apri il file .gitignore (se non esiste già, crealo: `touch .gitignore`)
    3. Aggiungi al file .gitignore una riga contenente il nome del file txt appena creato. 
    4. Controlla lo stato di Git <br>
        ```git status``` <br>
        Il file notes-locali.txt **non** dovrebbe comparire tra i file da aggiungere (dovrebbe comparire qui: `git status --ignored`)
    5. Commit/push solo il .gitignore
          - Attenzione: il file txt non va aggiunto.
          - Aggiungi solo .gitignore:
            ```
            git add .gitignore
            git commit -m "Add local notes file to gitignore"
            git push origin main
            ```

    6. Risultato finale

    Alla fine dovresti avere:

    - un file .txt presente solo in locale;
    - una regola dentro .gitignore;
    - un commit che carica solo .gitignore;
    - nessun file .txt su GitHub.
    
    
NB `.gitignore` funziona bene sui file **nuovi non ancora tracciati**. Se un file era già stato committato prima, ignorarlo non basta: va prima rimosso dal tracking con `git rm --cached nomefile`.
